# Clase Lampara
* con 3 atributos (tipo, potencia, material)
* 2 funciones leerDatos(), mostrarDatos()  
* 2 objetos (lampara1, lampara2).
#### que sea interactivo --> los input
#### no usar while true, if, for, lista y diccionario
![link text](https://raw.githubusercontent.com/SantiagoMedrano128/-ExaUII-Clase-Lampara-Santiago-0103/refs/heads/main/infoexaIIclassLampara.png)

```python
class Lampara:
  def __init__(self, tipo, potencia, material):
    self.tipo = tipo
    self.potencia = potencia
    self.material = material # Fixed a missing assignment here as well, material was not being set

  def leerDatos(self):
    self.tipo = input("Ingrese el tipo de lampara: ")
    self.potencia = input("Ingrese la potencia de la lampara: ")
    self.material = input("Ingrese el material de la lampara: ")

  def mostrarDatos(self):
    print("Tipo de Lampara:", self.tipo)
    print("Potencia de Lampara:", self.potencia)
    print("Material de Lampara:", self.material)

print("Lampara 1")
lampara1 = Lampara("", "", "")
lampara1.leerDatos()
lampara1.mostrarDatos()
print("*-*-*-*-*-*-*-*-*-*-*-")
print("Lampara 2")
lampara2 = Lampara("", "", "")
lampara2.leerDatos()
lampara2.mostrarDatos()
print("*-*-*-*-*-*-*-*-*-*-*-")
print("Programa realizad por: Santiago Medrano NL = 37")
```
